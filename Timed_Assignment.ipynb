```python
#--@1--

# list
numbers = [12, 7, 9, 4, 22, 15, 8, 3, 30, 11]

even= 0
odd= 0

print("Checking the nos. are even or odd")
for num in numbers:

  if num % 2 == 0:
        print(f"{num} is even")
        even+= 1
  else:
        print(f"{num} is odd")
        odd+= 1

        even_count = len([num for num in numbers if num % 2 == 0])
        odd_count = len([num for num in numbers if num % 2 != 0])

# List of only even nos. :)
even_nos_list= [n for n in numbers if not n % 2]



print("outputs")
print(f"total even nos.: {even_count}")
print(f"total odd nos.: {odd_count}")
print(f"list of only even nos.: {even_nos_list}")
```

```python
#@--2--

#function

def calculate_avg(numbers: list):
  return sum(numbers) / len(numbers)

#Trying with list
try:
  avg= calculate_avg([100,20,300])
  print("the avg is:",avg)
except ZeroDivisionError:
  print("cannot find the avg")

#Trying with a zero list
try:
  avg= calculate_avg([])
  print("the avg is:",avg)
except ZeroDivisionError:
  print("cannot find the avg")
```

```python
#--@3--

#function

import random
import math


guesses = []

#loop for random gusesses and printing

for n in range(10):
   guess = random.randint(1, 50)
   guesses.append(guess)
   print("Guess", n + 1, ":", guess)

#highest and average picks by random function..

highest = max(guesses)
average = sum(guesses) / len(guesses)

print(guesses)
print("Highest guess:", highest)
print("Average guess:", round(average, 2))
```

```python
#--@4--

#Class
class Student:
    def __init__(self, name: str, marks: list):
        self.name = name
        self.marks = marks

    def average(self):
        return sum(self.marks) / len(self.marks)

    def __str__(self):
        return f"{self.name}: {self.average():.2f}"


students = [
    Student("CHINMAY", [45, 33, 49]),
    Student("DATTA", [40, 36, 48]),
    Student("PRANIL", [46, 35, 47])
]

highest_student = students[0]

for student in students:
    if student.average() > highest_student.average():
        highest_student = student

print("Highest average:", highest_student.name)
```
